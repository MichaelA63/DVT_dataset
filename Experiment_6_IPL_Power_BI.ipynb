## Theory
Power BI is a Business Intelligence (BI) tool developed by Microsoft that enables users to import data, transform data, build dashboards, and perform business analytics. Power BI converts raw data into meaningful insights through interactive visualizations.

## Dataset
This experiment uses the supplied IPL ball-by-ball dataset.

## Learning Outcome
The IPL dataset can be transformed into an interactive dashboard containing KPI cards, charts, and slicers for season and team-wise analysis.


````python
import pandas as pd
import numpy as np

ipl = pd.read_csv("IPL_ball_by_ball_updated.csv")

print("Dataset loaded successfully.")
print("Rows:", len(ipl))
print("Columns:", len(ipl.columns))
print("Shape:", ipl.shape)
print("Seasons:", int(ipl["season"].min()), "to", int(ipl["season"].max()))
print("\nColumn names:")
print(list(ipl.columns))
````

## Algorithm 1 – Import Data into Power BI

1. Open **Power BI Desktop**.
2. Select **Home → Get Data → Text/CSV**.
3. Select `IPL_ball_by_ball_updated.csv`.
4. Preview the data.
5. Click **Load**.

The same IPL CSV used in this notebook is supplied with the experiment.


````python
print("First 10 rows of the IPL dataset:")
ipl.head(10)
````

## Algorithm 2 – Transform Data using Power Query

1. Select **Home → Transform Data**.
2. Check the data types of all columns.
3. Rename columns if required.
4. Check for null or blank values.
5. Remove unnecessary columns where appropriate.
6. Apply the transformations using **Close & Apply**.


````python
print("Data types:")
print(ipl.dtypes.to_string())

print("\nMissing values:")
print(ipl.isnull().sum().to_string())
````

## Algorithm 3 – Create KPI Measures

Create the following DAX measures in Power BI:

```DAX
Total Runs = SUM(IPL_ball_by_ball_updated[runs_off_bat])

Total Matches = DISTINCTCOUNT(IPL_ball_by_ball_updated[match_id])

Total Deliveries = COUNTROWS(IPL_ball_by_ball_updated)

Runs Per Match = DIVIDE([Total Runs], [Total Matches])
```

These measures can be displayed using **Card** visuals.


````python
total_runs = int(ipl["runs_off_bat"].sum())
total_matches = int(ipl["match_id"].nunique())
total_deliveries = int(len(ipl))
runs_per_match = round(total_runs / total_matches, 2)

print("KPI VALUES FROM THE SUPPLIED IPL DATA")
print("--------------------------------------")
print("Total Runs:", total_runs)
print("Total Matches:", total_matches)
print("Total Deliveries:", total_deliveries)
print("Runs Per Match:", runs_per_match)
````

## Algorithm 4 – Line Chart: Total Runs by Season

In Power BI:
- Insert **Line Chart**
- X-axis → `season`
- Y-axis → `Total Runs`
- Add a suitable title

This visualization shows the trend of IPL runs across seasons.


````python
season_summary = (
    ipl.groupby("season")
       .agg(
           Total_Runs=("runs_off_bat", "sum"),
           Matches=("match_id", "nunique")
       )
       .reset_index()
)

print("Season-wise total runs and matches:")
season_summary
````

## Algorithm 5 – Bar Chart: Team-wise Total Runs

In Power BI:
- Insert **Clustered Bar Chart**
- Axis → `batting_team`
- Values → `Total Runs`
- Sort by Total Runs in descending order

This compares the run contribution of IPL teams.


````python
team_summary = (
    ipl.groupby("batting_team")
       .agg(
           Total_Runs=("runs_off_bat", "sum"),
           Matches=("match_id", "nunique")
       )
       .reset_index()
       .sort_values("Total_Runs", ascending=False)
)

print("Team-wise total runs:")
team_summary
````

## Algorithm 6 – Pie/Donut Chart: Team Contribution

In Power BI:
- Insert **Donut Chart**
- Legend → `batting_team`
- Values → `Total Runs`

The chart shows the proportion of total IPL runs contributed by each team.


````python
team_share = team_summary[["batting_team", "Total_Runs"]].copy()
team_share["Percentage"] = (
    team_share["Total_Runs"] / team_share["Total_Runs"].sum() * 100
).round(2)

print("Team contribution percentages:")
team_share
````

## Algorithm 7 – Scatter Chart: Runs vs Deliveries

In Power BI:
- Insert **Scatter Chart**
- X-axis → number of deliveries
- Y-axis → total runs
- Details → `batting_team`

This can be used to study the relationship between team batting activity and runs.


````python
scatter_summary = (
    ipl.groupby("batting_team")
       .agg(
           Total_Runs=("runs_off_bat", "sum"),
           Deliveries=("match_id", "size")
       )
       .reset_index()
)

print("Team-wise runs and deliveries:")
print(scatter_summary.to_string(index=False))
print("\nCorrelation between team deliveries and total runs:",
      round(scatter_summary["Deliveries"].corr(scatter_summary["Total_Runs"]), 4))
````

## Algorithm 8 – Add Interactive Slicers

Add slicers in Power BI for:
- `season`
- `batting_team`
- `bowling_team`

Selecting a value in a slicer filters the dashboard visuals interactively.


````python
print("Available seasons:")
print(sorted(ipl["season"].dropna().unique().tolist()))

print("\nNumber of batting teams:", ipl["batting_team"].nunique())
print("Number of bowling teams:", ipl["bowling_team"].nunique())
````
