## Loading the IPL Dataset

The uploaded IPL dataset contains ball-by-ball records. The notebook uses the combined CSV supplied for this project.

```python
ipl = pd.read_csv("IPL_ball_by_ball_updated.csv")

print("IPL dataset loaded successfully.")
print("Rows:", ipl.shape[0])
print("Columns:", ipl.shape[1])
```

```python
print("First 5 records:")
display(ipl.head())
```

```python
print("Dataset columns:")
print(ipl.columns.tolist())
```

```python
print("Dataset data types:")
display(ipl.dtypes.to_frame("Data Type"))
```

```python
print("Dataset shape:", ipl.shape)
print("Number of seasons:", ipl["season"].nunique())
print("Seasons:", sorted(ipl["season"].unique()))
```

## Matplotlib Demonstration

Matplotlib is used to create charts and visualizations in Python.

```python
import matplotlib.pyplot as plt

season_runs = ipl.groupby("season")["runs_off_bat"].sum().sort_index()

plt.figure(figsize=(9, 5))
plt.plot(season_runs.index, season_runs.values, marker="o")
plt.title("IPL Total Runs by Season")
plt.xlabel("Season")
plt.ylabel("Runs Off Bat")
plt.xticks(season_runs.index, rotation=45)
plt.tight_layout()
plt.show()
```

```python
print("Visualization data:")
display(season_runs.to_frame("Total Runs"))
```
