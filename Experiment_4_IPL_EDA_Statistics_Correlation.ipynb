## 1. Import Dataset and Libraries

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

matches = pd.read_csv("matches.csv")
deliveries = pd.read_csv("deliveries.csv")

print("Matches shape:", matches.shape)
print("Deliveries shape:", deliveries.shape)
```

### Output

```python

# Sample output (exact values depend on your IPL dataset)
print(f"Matches shape: {matches.shape}")
print(f"Deliveries shape: {deliveries.shape}")
```

## Algorithm 1 – Dataset Inspection

1. Import dataset
2. Display dataset
3. Check dimensions
4. Display data types
5. Generate summary

```python

print("First 5 rows of matches:")
display(matches.head())

print("Dataset shape:")
print(matches.shape)

print("\nData types:")
print(matches.dtypes)

print("\nDescriptive summary:")
display(matches.describe(include="all").T)
```

### Output


**Output:** The notebook displays the first five IPL match records, the number of rows and columns, each column's data type, and a descriptive summary of the dataset.

> The exact rows, dimensions and statistics are generated from your local `matches.csv`.


## Algorithm 2 – Filtering Data

1. Specify condition
2. Apply filter
3. Display records

### Program 4 – IPL Matches by a Specific Team

```python

# Select a team dynamically from the available IPL team columns.
team_col = next(
    (c for c in ["team1", "Team1", "winner"] if c in matches.columns),
    None
)

if team_col:
    selected_team = matches[team_col].dropna().astype(str).iloc[0]
    filtered_matches = matches[
        matches[team_col].astype(str) == selected_team
    ]
    print(f"Column used: {team_col}")
    print(f"Team selected: {selected_team}")
    display(filtered_matches.head(10))
else:
    print("No suitable team column found.")
```

### Output

**Output:** Displays IPL match records satisfying the selected team condition. The team is taken automatically from the first available value in the dataset.

### Program 5 – Filter High-Scoring Deliveries

```python

run_col = next(
    (c for c in ["total_runs", "total_runs_y", "batsman_runs"] if c in deliveries.columns),
    None
)

if run_col:
    high_runs = deliveries[deliveries[run_col] > deliveries[run_col].median()]
    print(f"Column used: {run_col}")
    print(f"Median value: {deliveries[run_col].median()}")
    display(high_runs.head(10))
else:
    print("No suitable run column found.")
```

### Output

**Output:** Displays deliveries whose selected run value is greater than the median of that column.

## Algorithm 3 – Sorting Data

1. Select column
2. Apply `sort_values()`
3. Display sorted result

```python

if run_col:
    sorted_deliveries = deliveries.sort_values(
        by=run_col,
        ascending=False
    )
    display(sorted_deliveries[[run_col]].head(10))
else:
    print("No suitable numerical column found.")
```

### Output

**Output:** The selected IPL numerical column is displayed in descending order, showing the highest values first.

## Visualization – IPL Run Ranking

```python

if run_col:
    top_values = deliveries[run_col].value_counts().sort_index()

    plt.figure(figsize=(8, 4))
    top_values.plot(kind="bar")
    plt.title(f"Frequency of {run_col}")
    plt.xlabel(run_col)
    plt.ylabel("Frequency")
    plt.xticks(rotation=0)
    plt.show()
```

### Output

**Output:** A Matplotlib bar chart showing the frequency distribution of the selected IPL run variable.

## Algorithm 4 – Statistical Analysis

1. Select numerical column
2. Calculate Mean
3. Calculate Median
4. Calculate Mode
5. Calculate Standard Deviation

```python

if run_col:
    mean_value = deliveries[run_col].mean()
    median_value = deliveries[run_col].median()
    mode_value = deliveries[run_col].mode()
    std_value = deliveries[run_col].std()

    print("Mean =", mean_value)
    print("Median =", median_value)
    print("Mode =", mode_value.tolist())
    print("Standard Deviation =", std_value)
else:
    print("No numerical IPL column found.")
```

### Output

**Output:** Mean, median, mode and standard deviation of the selected IPL numerical column are printed.

### Important Formula

**Mean:** x̄ = Σx / n

```python

if run_col:
    print("Descriptive statistics:")
    display(deliveries[[run_col]].describe())
```

### Output

**Output:** Count, mean, standard deviation, minimum, quartiles and maximum are displayed.

## Algorithm 5 – Grouping and Aggregation

1. Select a categorical column
2. Apply `groupby()`
3. Calculate an average numerical value
4. Display result

```python

category_col = next(
    (c for c in ["winner", "team1", "team2", "venue", "city"] if c in matches.columns),
    None
)

numeric_match_col = next(
    (c for c in ["win_by_runs", "win_by_wickets", "target_runs", "result_margin"] if c in matches.columns),
    None
)

if category_col and numeric_match_col:
    group_avg = matches.groupby(category_col)[numeric_match_col].mean().sort_values(ascending=False)
    display(group_avg.head(10))
else:
    print("Suitable categorical/numerical columns for grouping were not found.")
```

### Output

**Output:** The average of the selected numerical match measure is calculated for each category, such as team, venue or city.

## Visualization – Category-wise Average

```python

if category_col and numeric_match_col:
    group_avg.head(10).plot(kind="bar", figsize=(9, 4))
    plt.title(f"Average {numeric_match_col} by {category_col}")
    plt.xlabel(category_col)
    plt.ylabel(f"Average {numeric_match_col}")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()
```

### Output

**Output:** A Matplotlib bar chart visualizes the category-wise average of the selected IPL numerical measure.

## Algorithm 6 – Correlation Analysis

1. Select numerical columns
2. Apply `corr()`
3. Interpret the correlation coefficient

```python

numeric_columns = deliveries.select_dtypes(include=np.number).columns.tolist()

# Select up to 8 numerical columns with useful variation.
usable_numeric = [
    c for c in numeric_columns
    if deliveries[c].nunique(dropna=True) > 1
][:8]

if len(usable_numeric) >= 2:
    corr = deliveries[usable_numeric].corr()
    display(corr)
else:
    print("At least two suitable numerical columns are required.")
```

### Output

**Output:** A correlation matrix is displayed for the selected numerical IPL variables. Values close to +1 indicate strong positive relationships, values close to -1 indicate strong negative relationships, and values close to 0 indicate weak or no linear relationship.

## Visualization – Correlation Matrix

```python

if len(usable_numeric) >= 2:
    plt.figure(figsize=(9, 7))
    plt.imshow(corr, aspect="auto")
    plt.colorbar(label="Correlation")
    plt.xticks(range(len(usable_numeric)), usable_numeric, rotation=45, ha="right")
    plt.yticks(range(len(usable_numeric)), usable_numeric)
    plt.title("IPL Numerical Variables – Correlation Matrix")
    plt.tight_layout()
    plt.show()
```

### Output

**Output:** A Matplotlib correlation-matrix visualization is generated to show relationships among IPL numerical variables.
