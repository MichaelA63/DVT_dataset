## 1. Import Dataset and Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("IPL_ball_by_ball_updated.csv")

print("Dataset shape:", df.shape)
print("Number of matches:", df["match_id"].nunique())
print("Seasons:", sorted(df["season"].dropna().unique().tolist()))
```

## 2. Data Inspection

```python
print("Columns:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())
```

## 3. Filtering and Sorting

```python
latest_season = df["season"].max()
filtered = df[(df["season"] == latest_season) & (df["runs_off_bat"] >= 4)]

print(f"Rows from season {latest_season} with 4 or more batter runs: {len(filtered)}")
display(filtered[["season", "match_id", "batting_team", "bowling_team", "runs_off_bat", "extras"]].head(10))

print("\nTop 10 deliveries by runs off the bat:")
display(df.sort_values("runs_off_bat", ascending=False)
        .head(10)[["season", "match_id", "batting_team", "bowling_team", "runs_off_bat", "extras"]])
```

## Algorithm 1 – Dataset Inspection
1. Import Pandas, NumPy and Matplotlib.
2. Read the IPL CSV dataset.
3. Display shape, columns, data types and missing values.
4. Inspect sample records using `head()`.

## Algorithm 2 – Filtering and Sorting
1. Select the latest season.
2. Filter deliveries with 4 or more batter runs.
3. Sort records by runs scored off the bat.
4. Display the selected records.


## 4. Grouping and Aggregation

```python
season_summary = (
    df.groupby("season")
      .agg(
          Total_Runs=("runs_off_bat", "sum"),
          Total_Deliveries=("match_id", "size"),
          Matches=("match_id", "nunique")
      )
      .reset_index()
)

print("Season-wise summary:")
display(season_summary)

team_summary = (
    df.groupby("batting_team")
      .agg(
          Total_Runs=("runs_off_bat", "sum"),
          Deliveries=("match_id", "size"),
          Average_Runs_Per_Delivery=("runs_off_bat", "mean")
      )
      .sort_values("Total_Runs", ascending=False)
)

print("Top 10 batting teams by total runs:")
display(team_summary.head(10))
```

## Algorithm 3 – Grouping and Aggregation
1. Group the dataset by season.
2. Calculate total runs, deliveries and matches.
3. Group the dataset by batting team.
4. Calculate total runs and average runs per delivery.
5. Sort the grouped result.


## 5. Descriptive Statistical Analysis

```python
runs = df["runs_off_bat"]

print("Descriptive statistics:")
display(runs.describe())

print("\nMean =", runs.mean())
print("Median =", runs.median())
print("Mode =", runs.mode().tolist())
print("Standard Deviation =", runs.std())
print("Variance =", runs.var())
print("Minimum =", runs.min())
print("Maximum =", runs.max())
```

## Algorithm 4 – Statistical Analysis
1. Select the numerical column `runs_off_bat`.
2. Calculate mean, median and mode.
3. Calculate standard deviation and variance.
4. Find minimum and maximum values.


## 6. Correlation Analysis

```python
numeric_columns = ["season", "innings", "runs_off_bat", "extras"]
corr = df[numeric_columns].corr()

print("Correlation matrix:")
display(corr)

print("Correlation between runs_off_bat and extras:",
      round(corr.loc["runs_off_bat", "extras"], 4))
```

## Algorithm 5 – Correlation
1. Select numerical variables from the IPL dataset.
2. Compute the Pearson correlation matrix using `corr()`.
3. Inspect relationships between the numerical variables.
4. Values close to +1 indicate a strong positive linear relationship, while values close to -1 indicate a strong negative relationship.


## 7. Visualization – Season-wise Total Runs

```python
plt.figure(figsize=(9, 5))
plt.plot(season_summary["season"], season_summary["Total_Runs"], marker="o")
plt.title("IPL Season-wise Total Runs")
plt.xlabel("Season")
plt.ylabel("Total Runs")
plt.grid(True)
plt.show()
```

## 8. Visualization – Top 10 Batting Teams

```python
top10 = team_summary.head(10).sort_values("Total_Runs")

plt.figure(figsize=(9, 5))
plt.barh(top10.index, top10["Total_Runs"])
plt.title("Top 10 IPL Batting Teams by Total Runs")
plt.xlabel("Total Runs")
plt.ylabel("Batting Team")
plt.tight_layout()
plt.show()
```

## 9. Visualization – Distribution of Runs per Delivery

```python
run_counts = df["runs_off_bat"].value_counts().sort_index()

plt.figure(figsize=(9, 5))
plt.bar(run_counts.index.astype(str), run_counts.values)
plt.title("Frequency Distribution of Runs Off the Bat")
plt.xlabel("Runs Off the Bat")
plt.ylabel("Number of Deliveries")
plt.show()
```

## 10. Visualization – Correlation Matrix

```python
plt.figure(figsize=(7, 5))
plt.imshow(corr, aspect="auto")
plt.colorbar(label="Correlation")
plt.xticks(range(len(numeric_columns)), numeric_columns, rotation=30, ha="right")
plt.yticks(range(len(numeric_columns)), numeric_columns)

for i in range(len(numeric_columns)):
    for j in range(len(numeric_columns)):
        plt.text(j, i, f"{corr.iloc[i, j]:.2f}", ha="center", va="center")

plt.title("IPL Numerical Variables – Correlation Matrix")
plt.tight_layout()
plt.show()
```
