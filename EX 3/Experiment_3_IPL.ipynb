# EDA – DATA INSPECTION AND ANALYSIS

**Dataset:** IPL Ball-by-Ball Dataset

### Aim
To inspect and analyze the IPL ball-by-ball dataset using Exploratory Data Analysis (EDA) to understand the data structure, identify patterns, detect missing values, and discover relationships between different IPL attributes.

### Procedure
1. Import Pandas, NumPy, Matplotlib, and Seaborn libraries.
2. Load the IPL ball-by-ball dataset using Pandas.
3. Display the first and last records using `head()` and `tail()`.
4. Check the number of rows and columns using `shape`.
5. Display column names and data types using `info()`.
6. Check missing values using `isnull().sum()`.
7. Generate statistical summaries using `describe()`.
8. Check and identify duplicate records.
9. Analyze important IPL attributes.
10. Create suitable graphs such as histograms, bar charts, and scatter plots.
11. Generate a correlation heatmap to identify relationships between numerical variables.
12. Interpret the findings and identify important patterns in the IPL dataset.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

file_path = "IPL_ball_by_ball_updated.csv"
df = pd.read_csv(file_path)

print("Dataset loaded successfully.")
```

## 1. Dataset Preview – First Records

```python
df.head()
```

## 2. Dataset Preview – Last Records

```python
df.tail()
```

## 3. Dataset Shape

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

## 4. Column Names and Data Types

```python
df.info()
```

## 5. Missing Values

```python
missing = df.isnull().sum()
print("Missing values in each column:")
print(missing)
```

## 6. Statistical Summary

```python
df.describe(include="all").transpose()
```

## 7. Duplicate Records

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

## 8. Analysis of Important IPL Attributes
The main attributes examined are season, batting team, bowling team, runs scored from the bat, extras, and wickets.

```python
print("Seasons:", sorted(df["season"].unique()))
print("\nNumber of batting teams:", df["batting_team"].nunique())
print("Number of bowling teams:", df["bowling_team"].nunique())
print("Total runs off bat:", df["runs_off_bat"].sum())
print("Total extras:", df["extras"].sum())
print("Total deliveries:", len(df))

wickets = df["wicket_type"].notna().sum()
print("Total wicket events:", wickets)
```

## 9. Histogram – Runs Scored Off the Bat

```python
plt.figure(figsize=(9,5))
plt.hist(df["runs_off_bat"], bins=np.arange(-0.5, 7.5, 1), edgecolor="black")
plt.xlabel("Runs Off the Bat")
plt.ylabel("Number of Deliveries")
plt.title("Distribution of Runs Off the Bat")
plt.xticks(range(7))
plt.tight_layout()
plt.show()
```

## 10. Bar Chart – Top Batting Teams by Total Runs

```python
team_runs = df.groupby("batting_team")["runs_off_bat"].sum().sort_values(ascending=False)
plt.figure(figsize=(11,6))
team_runs.head(10).sort_values().plot(kind="barh")
plt.xlabel("Total Runs Off the Bat")
plt.ylabel("Batting Team")
plt.title("Top 10 Batting Teams by Total Runs")
plt.tight_layout()
plt.show()

print(team_runs.head(10).to_string())
```

## 11. Scatter Plot – Ball Number vs Runs

```python
sample = df.sample(min(5000, len(df)), random_state=42)
plt.figure(figsize=(9,5))
plt.scatter(sample["ball"], sample["runs_off_bat"], alpha=0.25)
plt.xlabel("Ball Number in Over")
plt.ylabel("Runs Off the Bat")
plt.title("Ball Number vs Runs Off the Bat (Sample)")
plt.tight_layout()
plt.show()
```

## 12. Correlation Heatmap

```python
numeric_cols = ["season", "innings", "ball", "runs_off_bat", "extras"]
corr = df[numeric_cols].corr()

plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", square=True)
plt.title("Correlation Heatmap of Numerical IPL Attributes")
plt.tight_layout()
plt.show()
```

## 13. Findings and Interpretation

- The dataset contains ball-by-ball records covering multiple IPL seasons.
- `runs_off_bat` is concentrated at 0 runs, with boundaries forming smaller but important groups.
- Batting-team aggregation shows the teams with the highest cumulative runs across the available seasons.
- The scatter plot shows that runs are discrete delivery-level outcomes and do not have a strong direct relationship with the ball number within an over.
- The correlation heatmap summarizes linear relationships among the numerical attributes.
- Missing values are concentrated in event-specific columns such as wides, no-balls, wickets, and dismissals; these are naturally sparse because such events occur only on relevant deliveries.
- No duplicate records are present in the dataset.
