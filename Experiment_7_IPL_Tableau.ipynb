## Algorithm 3 – Bar Chart
**Objective:** Compare total runs scored by batting team.

### Tableau steps
- Put `batting_team` on **Rows**.
- Put `runs_off_bat` on **Columns** and use **SUM**.
- Sort descending.
- Enable data labels.
- Add a meaningful title.

````python
team_runs = df.groupby("batting_team")["runs_off_bat"].sum().sort_values(ascending=False)
print("Total runs by batting team:")
print(team_runs.to_string())
````

## Algorithm 4 – Line Chart
**Objective:** Show IPL batting-run trend across seasons.

### Tableau steps
- Put `season` on **Columns**.
- Put `runs_off_bat` on **Rows** using **SUM**.
- Select the **Line** mark.
- Add labels/tooltips and a suitable title.

````python
season_runs = df.groupby("season")["runs_off_bat"].sum().sort_index()
print("Total runs by season:")
print(season_runs.to_string())
````

## Algorithm 5 – Pie Chart
**Objective:** Show the share of total runs among the top batting teams.

### Tableau steps
- Put `batting_team` on **Color**.
- Put `SUM(runs_off_bat)` on **Angle**.
- Change Marks to **Pie**.
- Put team and run values on **Label**.
- Format labels as percentages where appropriate.

````python
top6 = team_runs.head(6)
share = (top6 / top6.sum() * 100).round(2)
print("Top 6 teams by total batting runs:")
display(pd.DataFrame({"Total Runs": top6, "Share (%)": share}))
````

## Algorithm 6 – Tables and Cross-Tabs
Create a season-versus-team cross-tab.

### Tableau steps
- `season` → **Rows**
- `batting_team` → **Columns**
- `runs_off_bat` → **Text**
- Aggregation → **SUM**

````python
cross_tab = pd.pivot_table(df, index="season", columns="batting_team",
                             values="runs_off_bat", aggfunc="sum", fill_value=0)
print("Season × Batting Team cross-tab:")
display(cross_tab)
````

## Algorithm 7 – Heatmap
Create a heatmap from the season/team cross-tab.

### Tableau steps
- `season` → Rows
- `batting_team` → Columns
- `SUM(runs_off_bat)` → Color
- Change Marks to **Square**
- Increase square size and format the color legend.

````python
print("Heatmap data prepared.")
print("Seasons:", cross_tab.shape[0])
print("Teams:", cross_tab.shape[1])
display(cross_tab.head())
````

## Algorithm 8 – Treemap

## Algorithm 11 – Calculated Fields
Create this Tableau calculated field:

**Runs Per Match**
```text
SUM([runs_off_bat]) / COUNTD([match_id])
```

This can be used in KPI cards, tooltips, tables, and dashboards.

````python
total_runs = int(df["runs_off_bat"].sum())
total_matches = int(df["match_id"].nunique())
runs_per_match = round(total_runs / total_matches, 2)

print("Total batsman runs:", total_runs)
print("Total matches:", total_matches)
print("Runs per match:", runs_per_match)
````
