## Theory

Data Import and Export is an important first step in Exploratory Data Analysis (EDA). Data may be stored in CSV, Excel, SQL, JSON, or web-based formats. Pandas provides functions such as `read_csv()`, `read_excel()`, `read_sql()`, `to_csv()`, `to_excel()`, and `to_json()` for working with these formats.

## Algorithm 1: Importing IPL CSV Files

1. Import the Pandas library.
2. Load the IPL CSV files using `read_csv()`.
3. Display the first few records.
4. Check the dataset structure.
5. Analyze the imported data.

```python
import pandas as pd

matches = pd.read_csv("matches.csv")
deliveries = pd.read_csv("deliveries.csv")

print("IPL CSV files imported successfully.")
```

## Program 1 – Reading the Matches CSV File

```python
print(matches)
```

## Program 2 – Reading the Deliveries CSV File

```python
print(deliveries)
```

## Program 3 – Display First Five Records

```python
print("First five records from Matches:")
display(matches.head())

print("First five records from Deliveries:")
display(deliveries.head())
```

## Program 4 – Display Dataset Information

```python
print("Matches Dataset Information:")
matches.info()

print("\nDeliveries Dataset Information:")
deliveries.info()
```

## Program 5 – Statistical Summary

```python
print("Statistical summary of Matches:")
display(matches.describe(include="all"))

print("Statistical summary of Deliveries:")
display(deliveries.describe(include="all"))
```

## Algorithm 2: Importing Excel File

1. Create or use an Excel file containing IPL match data.
2. Use `read_excel()` to load the file.
3. Display the records.
4. Verify the imported data.

```python
# Export a small IPL sample to Excel first, then read it back.
# This keeps the experiment self-contained.

matches_sample = matches.head(20)
matches_sample.to_excel("ipl_matches_sample.xlsx", index=False)

ipl_excel = pd.read_excel("ipl_matches_sample.xlsx")
display(ipl_excel.head())
```

## Algorithm 3: Importing SQL Data

1. Create a SQLite database.
2. Store IPL match records in a database table.
3. Retrieve records using an SQL query.

```python
import sqlite3

conn = sqlite3.connect("ipl_matches.db")

matches.head(20).to_sql("matches", conn, if_exists="replace", index=False)
query = "SELECT * FROM matches"
ipl_sql = pd.read_sql(query, conn)

display(ipl_sql.head())
conn.close()
```

## Data Export

Pandas supports exporting DataFrames into several formats:

| Format | Pandas Function |
|---|---|
| CSV | `to_csv()` |
| Excel | `to_excel()` |
| JSON | `to_json()` |
| SQL | `to_sql()` |

## Program 6 – Export IPL Matches to CSV

```python
matches.to_csv("ipl_matches_output.csv", index=False)
print("IPL Matches CSV File Exported")
```

## Program 7 – Export IPL Matches to Excel

```python
matches.to_excel("ipl_matches_output.xlsx", index=False)
print("IPL Matches Excel File Exported")
```

## Program 8 – Export IPL Matches to JSON

```python
matches.to_json("ipl_matches_output.json", orient="records", indent=2)
print("IPL Matches JSON File Exported")
```

## Program 9 – Export IPL Deliveries to CSV

```python
deliveries.head(1000).to_csv("ipl_deliveries_sample.csv", index=False)
print("IPL Deliveries Sample CSV File Exported")
```

## Basic Inspection of Imported IPL Data

```python
print("Matches shape:", matches.shape)
print("Deliveries shape:", deliveries.shape)

print("\nMatches columns:")
print(matches.columns.tolist())

print("\nDeliveries columns:")
print(deliveries.columns.tolist())
```
